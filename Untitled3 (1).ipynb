```python
# ==========================================
# SWYNEX Task 3: Visualization & Interpretation
# ==========================================

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# 1. Set the visual style for publication-ready figures
sns.set_theme(style="whitegrid")

# 2. Load or Prepare Dataset
# (If using your cleaned dataset from Task 2, uncomment the line below:)
# df = pd.read_csv('your_cleaned_dataset.csv')

# --- Simulated dataset for demonstration ---
np.random.seed(42)
data = {
    'Gene_ID': [f'Gene_{i}' for i in range(1, 101)],
    'Expression_Level': np.random.exponential(scale=50, size=100),
    'Biomarker_A': np.random.normal(loc=25, scale=5, size=100),
    'Clinical_Score': np.random.uniform(10, 100, size=100)
}
df = pd.DataFrame(data)
df['Clinical_Score'] = df['Biomarker_A'] * 2.5 + np.random.normal(0, 5, size=100)
# ------------------------------------------

# 3. Create Figure 1: Distribution Plot
plt.figure(figsize=(9, 5))
sns.histplot(df['Expression_Level'], kde=True, color='teal', bins=20)
plt.title('Distribution of Gene Expression Levels', fontsize=14, fontweight='bold')
plt.xlabel('Expression Level (FPKM)', fontsize=12)
plt.ylabel('Frequency (Number of Genes)', fontsize=12)
plt.savefig('expression_distribution.png', dpi=300, bbox_inches='tight')
plt.show()

# 4. Create Figure 2: Correlation Scatter Plot with Regression Line
plt.figure(figsize=(9, 5))
sns.regplot(
    data=df,
    x='Biomarker_A',
    y='Clinical_Score',
    color='purple',
    scatter_kws={'alpha': 0.7},
    line_kws={'color': 'red'}
)
plt.title('Correlation: Biomarker A Concentration vs. Clinical Score', fontsize=14, fontweight='bold')
plt.xlabel('Biomarker A Concentration', fontsize=12)
plt.ylabel('Clinical Score', fontsize=12)
plt.savefig('biomarker_correlation.png', dpi=300, bbox_inches='tight')
plt.show()

print("Charts successfully generated and saved as 'expression_distribution.png' and 'biomarker_correlation.png'!")
```
